```python
# Easting and Northing are coordinates of the British National Grid (OSGB36 coordinate system). The pyproj library can convert these to latitude and longitude coordinates. EPSG code: EPSG:27700. GPS: EPSG:4326 
# https://en.wikipedia.org/wiki/Ordnance_Survey_National_Grid
from pyproj import Transformer

def map_long_lat(df):
    transformer = Transformer.from_crs('epsg:27700', 'epsg:4326', always_xy=True)
    
    def parse_row(row):
        lon, lat = transformer.transform(row['Easting'], row['Northing'])
        return pd.Series({
            'Longitude': lon,
            'Latitude': lat
        })    
        
    return df.apply(parse_row, axis=1) # row wise

emissions_2019 = emissions_2019.join(map_long_lat(emissions_2019))
emissions_2019.head(2)
```

```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np

def plot_map(df, key_pollutant='co2', cluster_centers=None, fig_name='', plot_name=''):

    num_categories = df['Source Encoded'].nunique()  # count of unique sources
    distinct_colors = plt.get_cmap('viridis', num_categories)
    norm = mpl.colors.Normalize(vmin=df['Source Encoded'].min(), vmax=df['Source Encoded'].max())

    ax = df.plot(kind='scatter', x='Easting', y='Northing', grid=True, 
            s=df[key_pollutant] / 350, # size based on key emission type
            c=df['Source Encoded'], # colour based on emission source
            cmap=distinct_colors,
            alpha=0.5,
            colorbar=False,
            legend=False,
            figsize=(14,11))

    # colour bar legend
    cbar = plt.colorbar(mpl.cm.ScalarMappable(cmap=distinct_colors, norm=norm), ax=ax)
    cbar.set_ticks(np.arange(df['Source Encoded'].min(), df['Source Encoded'].max() + 1))
    unique_sources = df[['Source', 'Source Encoded']].drop_duplicates().sort_values('Source Encoded')
    cbar.set_ticks(unique_sources['Source Encoded'])
    cbar.set_ticklabels(unique_sources['Source'])
    cbar.set_label('Source')

    # size annotation
    plt.annotate(
        f'Relative size: {key_pollutant} emissions (tonnes/annum)', 
        xy=(0.98, 0.98),
        xycoords='axes fraction',
        ha='right', va='top',
        fontsize=10, color='black',
        bbox=dict(facecolor='white', edgecolor='blue', alpha=0.4)
    )


    if cluster_centers is not None:
        plt.plot(cluster_centers[:, 1],
                 cluster_centers[:, 0],
                 linestyle='', color='black', marker='X', markersize=20,
                 label='Cluster centers')

    london_img = plt.imread('./images/london.png')
    # axis = -0.6, 0.37, 51.23, 51.75 # GPS
    axis = 497846, 563695, 148799, 208357 # BNG

    plt.axis(axis)
    plt.imshow(london_img, extent=axis)

    # calculate the aspect ratio of the image, and adjust the plots aspect ratio to match.
    img_aspect_ratio = (axis[1] - axis[0]) / (axis[3] - axis[2])
    plt.gca().set_aspect(img_aspect_ratio, adjustable='box')

    plt.title(f'{key_pollutant} Emissions Scatter Plot {plot_name}')
    if fig_name != '':
        save_fig(fig_name)
    plt.show()

plot_map(emissions_2019)

# wood burning as an example
plot_map(emissions_2019[emissions_2019['Source'] == 'Wood Burning'])
produces_co2 = "No" if (emissions_2019[emissions_2019['Source'] == 'Wood Burning']['co2'] == 0).all() else "Yes"
print(f'Does Wood Burning produce Co2? -> {produces_co2}')
```

```python
high_emitters = emissions_2019[emissions_2019['co2'] > emissions_2019['co2'].quantile(0.999)]
plot_map(high_emitters)
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

kmeans = KMeans(1, n_init=5, random_state=42)
kmeans.fit_transform(high_emitters[['Northing', 'Easting']])
kmeans.cluster_centers_
```

```python
plot_map(high_emitters, cluster_centers=kmeans.cluster_centers_)
```

```python
emissions_2019['rank'] = emissions_2019.groupby('Source')['co2'].rank(method='first', ascending=False)
top_emitters_by_source = emissions_2019[emissions_2019['rank'] <= 20]

# take 5 sources we know produce high levels of Co2
for source in ['Commercial Shipping', 'Gas Combustion', 'Aviation', 'Part A1', 'NRMM Exhaust', 'HGV - Articulated']:
    top_emitters_for_source = top_emitters_by_source[top_emitters_by_source['Source'] == source]

    # and produce a cluster center for this source, 
    # weight them to push clusters closer to areas of high co2 volume
    kmeans = KMeans(1, n_init=5, random_state=42)
    kmeans.fit_transform(top_emitters_for_source[['Northing', 'Easting']], sample_weight=top_emitters_for_source['co2'])
    plot_map(top_emitters_for_source, cluster_centers=kmeans.cluster_centers_, plot_name=f'({source})')
```
